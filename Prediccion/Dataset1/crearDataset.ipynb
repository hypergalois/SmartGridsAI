```python
# Cargar dataset clima
path_clima = r"C:/Users/alexa/OneDrive/Documentos/Curso5/Proyectos4/Datos_clima/0200E_clima_numerizado.csv"
df_clima = pd.read_csv(path_clima)
```

```python
# Drop unnecessary columns
columns_to_drop = [
    'horatmin_sin', 'horatmin_cos', 'horatmax_minutos', 'horatmax_sin', 'horatmax_cos',
    'horaracha_minutos', 'horaracha_sin', 'horaracha_cos', 'horaPresMax_sin', 'horaPresMax_cos',
    'horaPresMin_sin', 'horaPresMin_cos', 'dir_sin', 'dir_cos'
]

df_clima.drop(columns=columns_to_drop, inplace=True)

#Columnas con las que nos quedamos
columnas_clima = ['fecha', 'altitud', 'tmed', 'tmin', 'tmax', 'velmedia', 'racha', 'sol', 'presMax', 'presMin',
                  'hrMedia', 'anio', 'mes', 'dia', 'dia_semana', 'estacion', 'horatmin_varias', 'horatmin_minutos',
                'horatmax_varias', 
                  'horaracha_varias','horaPresMax_varias',  'horaPresMin_varias', 
                'prec_inapreciable', 'prec_valor', 'prec_log']

df_clima = df_clima[columnas_clima]
```

```python
#cargar datasets de casas
casa_1 = pd.read_csv(r"C:/Users/alexa/OneDrive/Documentos/Curso5/Proyectos4/casa_12447.csv", parse_dates=['timestamp'])
casa_2 = pd.read_csv(r"C:/Users/alexa/OneDrive/Documentos/Curso5/Proyectos4/casa_7.csv", parse_dates=['timestamp'])
casa_3 = pd.read_csv(r"C:/Users/alexa/OneDrive/Documentos/Curso5/Proyectos4/casa_60.csv", parse_dates=['timestamp'])
casa_5 = pd.read_csv(r"C:/Users/alexa/OneDrive/Documentos/Curso5/Proyectos4/casa_5.csv", parse_dates=['timestamp'])
```

```python
# Ver los tipos de datos de cada columna en df_consumo
casa_1.drop(columns=['fecha'], inplace=True)
casa_2.drop(columns=['fecha'], inplace=True)
casa_3.drop(columns=['fecha'], inplace=True)
casa_5.drop(columns=['fecha'], inplace=True)
```

```python


# Unificar datos horarios de casa_1 con datos diarios de clima

# Convertir la columna 'fecha' del DataFrame clima a tipo datetime
df_clima['fecha'] = pd.to_datetime(df_clima['fecha'])

# Crear un rango de horas para cada día en el DataFrame clima
horas = pd.date_range(start='00:00', end='23:00', freq='H').time

# Expandir el DataFrame clima duplicando las filas para cada hora
df_clima_expandido = df_clima.loc[df_clima.index.repeat(len(horas))].copy()
df_clima_expandido['hora'] = list(horas) * len(df_clima)

# Crear una columna timestamp combinando fecha y hora
df_clima_expandido['timestamp'] = pd.to_datetime(df_clima_expandido['fecha'].astype(str) + ' ' + df_clima_expandido['hora'].astype(str))

# Unir el DataFrame casa_1 con el DataFrame clima_expandido en base al timestamp
df_casa_1_clima = pd.merge(casa_1, df_clima_expandido, on='timestamp', how='left')
df_casa_2_clima = pd.merge(casa_2, df_clima_expandido, on='timestamp', how='left')
df_casa_3_clima = pd.merge(casa_3, df_clima_expandido, on='timestamp', how='left')
df_casa_5_clima = pd.merge(casa_5, df_clima_expandido, on='timestamp', how='left')
```

```python
#hrMedia hay valores nulos quiero coger el valor de la fila anterior
df_casa_3_clima['hrMedia'] = df_casa_3_clima['hrMedia'].fillna(method='ffill')
```
