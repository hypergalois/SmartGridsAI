# N-HITS con Dart

### Importación de librerías y funciones

```python
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from darts import TimeSeries, concatenate
from darts.dataprocessing.transformers import MissingValuesFiller, Scaler
from darts.datasets import EnergyDataset
from darts.metrics import r2_score
from darts.models import NBEATSModel, NHiTSModel
from darts.utils.callbacks import TFMProgressBar
from darts.utils.timeseries_generation import datetime_attribute_timeseries


warnings.filterwarnings("ignore")
import logging

logging.disable(logging.CRITICAL)


def generate_torch_kwargs():
    return {
        "pl_trainer_kwargs": {
            "accelerator": "cpu",
            "callbacks": [TFMProgressBar(enable_train_bar_only=True)],
        }
    }
```

### Carga y visualización del conjunto de datos de energía

Se carga y visualiza una serie temporal de datos de generación hidroeléctrica.



```python
df = EnergyDataset().load().pd_dataframe()
df["generation hydro run-of-river and poundage"].plot()
plt.title("Hourly generation hydro run-of-river and poundage")
```

### Preprocesamiento de los datos y escalado



```python
#Agrupamos por día y calculamos la media
df_day_avg = df.groupby(df.index.astype(str).str.split(" ").str[0]).mean().reset_index()

#Rellenamos los valores faltantes
filler = MissingValuesFiller()

#Escalamos los datos
scaler = Scaler()

#Transformamos los datos a TimeSeries
series = filler.transform(
    TimeSeries.from_dataframe(
        df_day_avg, "time", ["generation hydro run-of-river and poundage"]
    )
).astype(np.float32)

#Dividimos los datos en train y val
train, val = series.split_after(pd.Timestamp("20170901"))

#Escalamos los datos
train_scaled = scaler.fit_transform(train)
val_scaled = scaler.transform(val)
series_scaled = scaler.transform(series)

#Visualizamos los datos
train_scaled.plot(label="training")
val_scaled.plot(label="val")
plt.title("Daily generation hydro run-of-river and poundage")
```

### Generación y escalado de covariables de fecha y hora

```python
# Generamos las covariables relacionadas con el año y el mes a partir de la serie original
hydro_year = datetime_attribute_timeseries(series, attribute='year')
hydro_month = datetime_attribute_timeseries(series, attribute='month')

# Apilamos las covariables de año y mes para formar una sola serie temporal
hydro_covariates = hydro_year.stack(hydro_month)

# Dividimos en train y val
nValLen = len(val_scaled)
hydro_train_covariates, hydro_val_covariates = hydro_covariates[:-nValLen], hydro_covariates[-nValLen:]

# Escalamos las covariables
scaler_covariates = Scaler()
hydro_train_covariates = scaler_covariates.fit_transform(hydro_train_covariates).astype(np.float32)
hydro_val_covariates = scaler_covariates.transform(hydro_val_covariates).astype(np.float32)

# Concatenamos las covariables escaladas de entrenamiento y validación para formar una serie completa de covariables
hydro_covariates = concatenate([hydro_train_covariates, hydro_val_covariates])

#Visualizamos las covariables
plt.figure()
hydro_covariates.plot()
plt.title('Hydro production covariates')
plt.show()
```

### Configuración y Entrenamiento del Modelo NHiTS

```python
model_name = "NHiTS"
model = NHiTSModel(
    input_chunk_length=30,            # Longitud de la secuencia de entrada (30 pasos de tiempo)
    output_chunk_length=7,            # Longitud de la secuencia de salida (7 días de predicción)
    num_stacks=10,                    # Número de pilas (stacks) en el modelo
    num_blocks=1,                     # Número de bloques dentro de cada pila
    num_layers=4,                     # Número de capas dentro de cada bloque
    layer_widths=512,                 # Ancho de cada capa (número de unidades)
    n_epochs=100,                     # Número de épocas para el entrenamiento
    nr_epochs_val_period=1,           # Frecuencia de evaluación del modelo en el conjunto de validación
    batch_size=800,                   # Tamaño de cada lote de entrenamiento
    random_state=42,                  # Semilla para garantizar reproducibilidad
    model_name=model_name,            # Nombre del modelo para guardar y cargar checkpoints
    save_checkpoints=True,            # Si se guarda el modelo en puntos de control durante el entrenamiento
    force_reset=True,                 # Si se restablece el modelo si ya ha sido entrenado previamente
    **generate_torch_kwargs()         
)


#Podemos elegir entre entrenar con covariables o sin ellas o con solo algunas ( puede cambiar el resultado de la prediccion y no siempre es mejor)
model.fit(series=[train_scaled], 
          #past_covariates=[hydro_train_covariates],
          val_series=[val_scaled],
          #val_past_covariates=[hydro_val_covariates])
            )
```

```python
# Cargar el modelo NHiTS previamente entrenado desde el punto de control más reciente y el mejor modelo
model_nhits = NHiTSModel.load_from_checkpoint(model_name=model_name, best=True)
```

### Predicción y Evaluación del Modelo NHiTS

```python
pred_series = model_nhits.historical_forecasts(
    series_scaled,
    start=pd.Timestamp("2017-09-02"),   # Fecha de inicio desde la cual hacer las predicciones
    forecast_horizon=7,
    stride=7,
    last_points_only=False,
    retrain=False,
    verbose=True,
    #past_covariates=[hydro_covariates]
)

pred_series = concatenate(pred_series)
```

```python
# Calcular el R² (coeficiente de determinación) entre la serie de validación y la serie de predicción (cuanto más cerca de 1, mejor)

r2 = r2_score(val_scaled, pred_series)
plt.figure(figsize=(10, 6))
series_scaled.plot(label="Actual Series")
pred_series.plot(label="Forecast")
plt.title(f"R2: {r2}")
plt.legend()
plt.show()
```
