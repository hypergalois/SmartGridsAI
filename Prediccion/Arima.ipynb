```python
#!/usr/bin/env python
"""
Script: arima_forecast.py
Descripción: Forecast del consumo local utilizando ARIMA (pmdarima).
"""

import pandas as pd
import matplotlib.pyplot as plt
import pmdarima as pm

# Cargar el dataset (asegúrate de que 'mock_consumo_local.csv' esté en el directorio)
df = pd.read_csv("mock_consumo_local.csv", parse_dates=["timestamp"])

# Filtrar datos para una única casa (por ejemplo, casa_id = 1)
df_house = df[df["casa_id"] == 1].copy()
df_house.sort_values("timestamp", inplace=True)
df_house.set_index("timestamp", inplace=True)

# Ajustar el modelo ARIMA, considerando estacionalidad diaria (m=24)
model_arima = pm.auto_arima(
    df_house["consumo_kW"],
    seasonal=True, 
    m=24,
    trace=True,
    error_action="ignore", 
    suppress_warnings=True
)

# Predecir las próximas 24 horas
n_periods = 24
forecast_arima = model_arima.predict(n_periods=n_periods)
forecast_index = pd.date_range(start=df_house.index[-1] + pd.Timedelta(hours=1), periods=n_periods, freq="H")
forecast_series = pd.Series(forecast_arima, index=forecast_index)

# Graficar el histórico y el forecast
plt.figure(figsize=(12, 6))
plt.plot(df_house["consumo_kW"][-100:], label="Histórico")
plt.plot(forecast_series, label="Forecast ARIMA", color="red")
plt.title("Forecast ARIMA - Consumo kW (Casa 1)")
plt.xlabel("Timestamp")
plt.ylabel("Consumo (kW)")
plt.legend()
plt.show()
```
