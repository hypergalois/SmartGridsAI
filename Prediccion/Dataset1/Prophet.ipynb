# Preproceso

```python
df['datetime'] = pd.to_datetime(df['datetime'])
# Ordenamos 
df = df.sort_values(by=['id_casa', 'datetime'])

```

```python
# Asegúrate de que la columna 'datetime' sea de tipo datetime
df['datetime'] = pd.to_datetime(df['datetime'])

def interpolate_humedad_por_casa(grupo):
    grupo = grupo.sort_values('datetime').copy()
    grupo = grupo.set_index('datetime')
    grupo['humedad'] = grupo['humedad'].interpolate(method='time')
    # Rellenar los valores nulos restantes utilizando forward fill y backward fill
    grupo['humedad'] = grupo['humedad'].ffill().bfill()
    grupo = grupo.reset_index()
    return grupo

df = df.groupby('id_casa', group_keys=False).apply(interpolate_humedad_por_casa)
```
