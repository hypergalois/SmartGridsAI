```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
dataset_produccion = pd.read_csv('datasets/dataset_produccion.csv')
dataset_consumo = pd.read_csv('datasets/dataset_consumo.csv')
```

```python
# Guardar las primeras 100 casas random
casas = dataset_produccion['id_casa'].unique()
len(casas)
```

```python
dataset_consumo = dataset_consumo[dataset_consumo['id_casa'].isin(casas[:100])]
```

```python
# graficar produccion de una sola casa
# Filtrar los datos para una sola casa (reemplaza 'id_casa' y 'produccion' con los nombres reales de las columnas)
id_casa_seleccionada = 3234  # Selecciona la primera casa del array 'casas'
datos_casa = dataset_produccion[dataset_produccion['id_casa'] == id_casa_seleccionada]

# Graficar la producción
plt.figure(figsize=(10, 6))
plt.plot(datos_casa['datetime'], datos_casa['produccion_kWh'], label=f'Casa {id_casa_seleccionada}')
plt.xlabel('Tiempo')  # Reemplaza con la unidad de tiempo real
plt.ylabel('Producción')
plt.title(f'Producción de la Casa {id_casa_seleccionada}')
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Graficar la nubosidad
plt.figure(figsize=(10, 6))
plt.plot(dataset_produccion['datetime'], dataset_produccion['nubosidad'], label='Nubosidad', color='blue')
plt.xlabel('Tiempo')  # Reemplaza con la unidad de tiempo real
plt.ylabel('Nubosidad')
plt.title('Nubosidad a lo largo del tiempo')
plt.legend()
plt.grid()
plt.show()  
```

```python
# Supongamos que dataset_consumo es un DataFrame de pandas
dataset_consumo.to_csv('datasets/dataset_consumo.csv', index=False)
dataset_produccion.to_csv('datasets/dataset_produccion.csv', index=False)
```
