```python
#!/usr/bin/env python
"""
Script: prophet_forecast.py
Descripción: Forecast del consumo local utilizando Prophet.
"""

import pandas as pd
import matplotlib.pyplot as plt
from prophet import Prophet

# Cargar el dataset
df = pd.read_csv("mock_consumo_local.csv", parse_dates=["timestamp"])

# Filtrar datos para una única casa (por ejemplo, casa_id = 1)
df_house = df[df["casa_id"] == 1].copy()
df_house.sort_values("timestamp", inplace=True)

# Preparar el DataFrame para Prophet (columnas 'ds' y 'y')
df_prophet = df_house[["timestamp", "consumo_kW"]].rename(columns={"timestamp": "ds", "consumo_kW": "y"})

# Inicializar y ajustar el modelo (con estacionalidad diaria activada)
model_prophet = Prophet(daily_seasonality=True)
model_prophet.fit(df_prophet)

# Crear un DataFrame futuro para predecir las próximas 24 horas
n_periods = 24
future = model_prophet.make_future_dataframe(periods=n_periods, freq="H")
forecast = model_prophet.predict(future)

# Graficar el forecast
fig = model_prophet.plot(forecast)
plt.title("Forecast Prophet - Consumo kW (Casa 1)")
plt.xlabel("Timestamp")
plt.ylabel("Consumo (kW)")
plt.show()
```
