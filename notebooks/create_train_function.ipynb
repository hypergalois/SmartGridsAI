```python
import mlrun

# Obtener el proyecto
project = mlrun.get_or_create_project("smartgrids")
```

```python
fn = project.set_function(
    func="train-model.py",
    name="train-forecasting-model",
    kind="job"
)

fn.deploy()
```

```python
# Ejecutar la función con el dataset preprocesado
run = project.run_function(
    "train-forecasting-model",
    inputs={"dataset_uri": "store://datasets/smartgrids/preprocess-data-preprocess-data_consumo_electrico_preprocesado#0:latest"},
    handler="train_forecasting_model"
)
```
