```python
#!/usr/bin/env python
"""
Script: nhits_forecast.py
Descripción: Forecast del consumo local utilizando el modelo NHITS de Darts.
"""

import pandas as pd
import matplotlib.pyplot as plt
from darts import TimeSeries
from darts.models import NHITS

# Cargar el dataset
df = pd.read_csv("mock_consumo_local.csv", parse_dates=["timestamp"])

# Filtrar datos para una única casa (por ejemplo, casa_id = 1)
df_house = df[df["casa_id"] == 1].copy()
df_house.sort_values("timestamp", inplace=True)

# Convertir el DataFrame a un objeto TimeSeries de Darts
series = TimeSeries.from_dataframe(df_house, "timestamp", "consumo_kW")

# Dividir la serie en entrenamiento y validación (80% - 20%)
train, val = series.split_before(0.8)

# Configurar el modelo NHITS
model_nhits = NHITS(input_chunk_length=24, output_chunk_length=24, n_epochs=100, random_state=42)

# Ajustar el modelo en el conjunto de entrenamiento
model_nhits.fit(train)

# Predecir las próximas 24 horas
forecast_nhits = model_nhits.predict(n=24)

# Graficar el histórico y el forecast
plt.figure(figsize=(12, 6))
series[-100:].plot(label="Histórico")
forecast_nhits.plot(label="Forecast NHITS", lw=2, ls="--")
plt.title("Forecast NHITS - Consumo kW (Casa 1)")
plt.xlabel("Timestamp")
plt.ylabel("Consumo (kW)")
plt.legend()
plt.show()
```
