```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset  
data = pd.read_csv('dataset_d3_limpio.csv')
```

```python
data.head()
```

```python
len(data['id_casa'].unique())
```

```python
data = data[data['num_placas'] != 0]
```

```python
# Guardar las primeras 100 casas random
casas = data['id_casa'].unique()
casas = np.random.choice(casas, 100, replace=False)
```

```python
len(data['id_casa'].unique())
```

```python
casas
```

```python
# Filtrar el DataFrame para mantener solo las filas con las casas seleccionadas
data_filtrado = data[data['id_casa'].isin(casas)]
```

```python
len(data_filtrado['id_casa'].unique())
```

```python
# Guardar el DataFrame filtrado en un archivo csv
data_filtrado.to_csv('dataset_d3_filtrado.csv', index=False)
```
